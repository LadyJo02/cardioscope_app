```python
import tensorflow as tf

def inspect_tflite_model(model_path):
    """
    Loads a TFLite model and prints its input and output tensor details.
    """
    try:
        # Load the TFLite model and allocate tensors.
        interpreter = tf.lite.Interpreter(model_path=model_path)
        interpreter.allocate_tensors()

        print(f"---  inspecting Model: {model_path} ---")

        # Get input tensor details.
        input_details = interpreter.get_input_details()
        print("\n[Input Details]")
        for i, detail in enumerate(input_details):
            print(f"  Input #{i}:")
            print(f"    Name: {detail['name']}")
            print(f"    Shape: {detail['shape']}")
            print(f"    Data Type: {detail['dtype']}")
            print(f"    Quantization: {detail['quantization']}")

        # Get output tensor details.
        output_details = interpreter.get_output_details()
        print("\n[Output Details]")
        for i, detail in enumerate(output_details):
            print(f"  Output #{i}:")
            print(f"    Name: {detail['name']}")
            print(f"    Shape: {detail['shape']}")
            print(f"    Data Type: {detail['dtype']}")
            print(f"    Quantization: {detail['quantization']}")
        
        print("------------------------------------------\n")

    except Exception as e:
        print(f"Error inspecting {model_path}: {e}")

# --- Main execution ---
if __name__ == "__main__":
    # Define paths to your models
    preprocess_model = 'preprocess_mel.tflite'
    classifier_model = 'tcn_snn_full.tflite'

    # Inspect both models
    inspect_tflite_model(preprocess_model)
    inspect_tflite_model(classifier_model)
```

```python
import os
os.getcwd()

import tensorflow as tf
from pathlib import Path

def inspect_tflite_model(model_path: str):
    """Inspect the input/output tensors of a TFLite model."""
    if not Path(model_path).exists():
        print(f"❌ File not found: {model_path}")
        return

    print(f"\n🔍 --- Inspecting Model: {model_path} ---")
    try:
        interpreter = tf.lite.Interpreter(model_path=model_path)
        interpreter.allocate_tensors()

        # --- Input Details ---
        input_details = interpreter.get_input_details()
        print("\n[Input Details]")
        for i, d in enumerate(input_details):
            print(f"  • Input #{i}")
            print(f"    Name         : {d['name']}")
            print(f"    Shape        : {d['shape']}")
            print(f"    Data Type    : {d['dtype']}")
            print(f"    Quantization : {d['quantization']}")

        # --- Output Details ---
        output_details = interpreter.get_output_details()
        print("\n[Output Details]")
        for i, d in enumerate(output_details):
            print(f"  • Output #{i}")
            print(f"    Name         : {d['name']}")
            print(f"    Shape        : {d['shape']}")
            print(f"    Data Type    : {d['dtype']}")
            print(f"    Quantization : {d['quantization']}")

    except Exception as e:
        print(f"⚠️ Error inspecting {model_path}: {e}")
    print("---------------------------------------------")


if __name__ == "__main__":
    # 🔸 Update these paths to your actual local model files
    models = [
        "preprocess_mel.tflite",
        "tcn_snn_full.tflite",
        "lstm_full.tflite",
    ]
    
    for m in models:
        inspect_tflite_model(m)
```
